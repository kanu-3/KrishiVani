```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("processed/tomato_model_ready.csv")
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df = df.sort_values("date").reset_index(drop=True)
```

```python
print(df.head())
print(df.shape)
print(df.isna().sum())
```

```python
df["year"] = df["date"].dt.year
```

```python
df["month"] = df["date"].dt.month
```

```python
df["day"] = df["date"].dt.day
```

```python
df["day_of_week"] = df["date"].dt.dayofweek
```

```python
df["day_of_year"] = df["date"].dt.dayofyear
```

```python
df["lag_1"] = df["price"].shift(1)
df["lag_7"] = df["price"].shift(7)
df["lag_14"] = df["price"].shift(14)
df["lag_30"] = df["price"].shift(30)
```

```python
df.head()
```

```python
df["rolling_7_mean"] = (
    df["price"]
    .rolling(window=7)
    .mean()
)
```

```python
df["rolling_14_mean"] = (
    df["price"]
    .rolling(window=14)
    .mean()
)
```

```python
df["rolling_30_mean"] = (
    df["price"]
    .rolling(window=30)
    .mean()
)
```

```python
df["rolling_7_std"] = (
    df["price"]
    .rolling(window=7)
    .std()
)
```

```python
df["rolling_30_std"] = (
    df["price"]
    .rolling(window=30)
    .std()
)
```

```python
df["lag_365"] = df["price"].shift(365)
```

```python
print(df.head(50))
```

```python
print(df.columns.tolist())
```

```python
df["target"] = df["price"].shift(-1)
```

```python
print(df.isna().sum())
```

```python
print("\nRows with missing price:")
print(
    df[df["price"].isna()][
        ["date", "price", "was_imputed"]
    ].to_string(index=False)
)
```

```python
print("\nRows with missing lag_365:")
print(
    df[df["lag_365"].isna()][
        ["date", "price", "lag_365"]
    ].head(20)
)
```

```python
print("\nMissing values by column:")
print(df.isna().sum())
```

```python
feature_columns = [
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_30",
    "rolling_7_mean",
    "rolling_14_mean",
    "rolling_30_mean",
    "rolling_7_std",
    "rolling_30_std",
    "month",
    "day_of_week",
    "day_of_year",
]

target_column = "target"

model_df = df.dropna(
    subset=feature_columns + [target_column]
).copy()

print("Original rows:", len(df))
print("Model-ready rows:", len(model_df))
print("Rows removed:", len(df) - len(model_df))

print("\nRemaining missing values:")
print(model_df[feature_columns + [target_column]].isna().sum())

print("\nModel-ready date range:")
print(model_df["date"].min())
print(model_df["date"].max())
```

```python
model_df = model_df.sort_values("date").reset_index(drop=True)
```

```python
train = model_df[model_df["date"].dt.year <= 2023].copy()
validation = model_df[model_df["date"].dt.year == 2024].copy()
test = model_df[model_df["date"].dt.year == 2025].copy()
```

```python
print("TRAIN")
print(train["date"].min(), "→", train["date"].max())
print("Rows:", len(train))
```

```python
print("\nVALIDATION")
print(validation["date"].min(), "→", validation["date"].max())
print("Rows:", len(validation))
```

```python
print("\nTEST")
print(test["date"].min(), "→", test["date"].max())
print("Rows:", len(test))
```

```python
features = [
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_30",
    "rolling_7_mean",
    "rolling_14_mean",
    "rolling_30_mean",
    "rolling_7_std",
    "rolling_30_std",
    "month",
    "day_of_week",
    "day_of_year",
]
```

```python
target = "target"
```

```python
X_train = train[features]
y_train = train[target]

X_val = validation[features]
y_val = validation[target]

X_test = test[features]
y_test = test[target]
```

```python
print("TRAIN:", len(train))
print("VALIDATION:", len(validation))
print("TEST:", len(test))
```
